```python

# Create an empty list to store all ticket records
records = []

# Create a random number generator with a fixed seed
rng = random.Random(42)

# Start ticket numbering from 1
ticket_number = 1

# Go through each category and its tickets
for category, tickets in tickets_by_category.items():

    # Copy the tickets so the original dictionary stays unchanged
    shuffled_tickets = tickets.copy()

    # Shuffle tickets to make the split less order-dependent
    rng.shuffle(shuffled_tickets)

    # Give each ticket its own ID and split
    for index, ticket in enumerate(shuffled_tickets):

        # First 8 tickets go to development, remaining 2 to evaluation
        if index < 8:
            split = "development"
        else:
            split = "eval"

        # Create one record
        records.append({
            "id": f"T{ticket_number:03d}",
            "ticket": ticket,
            "expected_category": category,
            "split": split
        })

        # Increase ticket number
        ticket_number += 1

# Convert records into a pandas DataFrame
df = pd.DataFrame(records)

# Display the first five rows
display(df.head())
```

```python

# Create a dictionary containing keywords for each category

keyword_rules = {

    "Login & Account Access": [
        "login", "log in", "password", "username",
        "account", "sign in", "verification code"
    ],

    "Wi-Fi & Network": [
        "wifi", "wi-fi", "internet", "network",
        "connection", "disconnecting", "wireless"
    ],

    "Software Installation": [
        "install", "installation", "software",
        "update", "updating", "application", "setup"
    ],

    "Hardware Problem": [
        "laptop", "keyboard", "battery", "screen",
        "monitor", "computer", "touchpad", "usb ports"
    ],

    "Email Issue": [
        "email", "inbox", "outbox", "messages",
        "send emails", "receiving emails", "attachments"
    ],

    "LMS & Portal Issue": [
        "lms", "portal", "course page", "online quiz",
        "timetable", "attendance", "course materials"
    ],

    "Printer Issue": [
        "printer", "printing", "print", "paper jam",
        "printer queue", "blank pages"
    ],

    "Data & File Access": [
        "file", "files", "folder", "shared drive",
        "document", "deleted", "cloud storage", "access denied"
    ],

    "Security & Phishing": [
        "suspicious", "phishing", "virus", "antivirus",
        "unauthorized", "compromised", "harmful",
        "security", "unfamiliar login"
    ],

    "General IT Support": [
        "projector", "workstation", "technical assistance",
        "computer facilities", "it services",
        "technology facilities", "technical question"
    ]
}

print("Keyword rules created successfully!")
print("Total categories:", len(keyword_rules))
```

```python
def keyword_baseline(ticket):
    ticket = ticket.lower()

    # Give priority to security-related tickets
    security_keywords = [
        "suspicious", "phishing", "unauthorized",
        "compromised", "virus", "harmful"
    ]

    for keyword in security_keywords:
        if keyword in ticket:
            return "Security & Phishing"

    # Count keyword matches for all categories
    scores = {}

    for category, keywords in keyword_rules.items():
        score = 0

        for keyword in keywords:
            if keyword in ticket:
                score += 1

        scores[category] = score

    best_category = max(scores, key=scores.get)

    if scores[best_category] == 0:
        return "General IT Support"

    return best_category
```

```python
# Get ticket text from development dataset
development_tickets = development_df["ticket"]

# Create TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Convert tickets into numerical vectors
development_vectors = tfidf_vectorizer.fit_transform(
    development_tickets
)

# Display the results
print("TF-IDF conversion completed!")
print("Number of tickets:", development_vectors.shape[0])
print("Number of features:", development_vectors.shape[1])
```

```python
# Select one ticket from evaluation dataset
new_ticket = evaluation_df.iloc[0]["ticket"]

# Convert the new ticket into a TF-IDF vector
new_ticket_vector = tfidf_vectorizer.transform([new_ticket])

# Compare it with all development tickets
similarity_scores = cosine_similarity(
    new_ticket_vector,
    development_vectors
)

# Find the most similar development ticket
best_match_index = similarity_scores.argmax()

# Get the category of the most similar ticket
predicted_category = development_df.iloc[
    best_match_index
]["expected_category"]

# Display results
print("New Ticket:", new_ticket)
print("Predicted Category:", predicted_category)
print("Similarity Score:", similarity_scores[0][best_match_index])
```

```python
def tfidf_predict(ticket):

    # Convert ticket into TF-IDF vector
    ticket_vector = tfidf_vectorizer.transform([ticket])

    # Calculate similarity with development tickets
    similarity_scores = cosine_similarity(
        ticket_vector,
        development_vectors
    )

    # Find the most similar ticket
    best_match_index = similarity_scores.argmax()

    # Get the predicted category
    predicted_category = development_df.iloc[
        best_match_index
    ]["expected_category"]

    return predicted_category
```

```python
from typing import Literal
from pydantic import BaseModel, Field

class TicketResponse(BaseModel):

    category: Literal[
        "Login & Account Access",
        "Wi-Fi & Network",
        "Software Installation",
        "Hardware Problem",
        "Email Issue",
        "LMS & Portal Issue",
        "Printer Issue",
        "Data & File Access",
        "Security & Phishing",
        "General IT Support"
    ]

    summary: str = Field(min_length=5, max_length=200)
```

```python
# Create a valid ticket response
valid_response = {
    "category": "Wi-Fi & Network",
    "summary": "The student's campus Wi-Fi keeps disconnecting."
}

# Validate the response using Pydantic
try:
    result = TicketResponse(**valid_response)
    print("Valid response!")
    print(result.model_dump())

except ValidationError as e:
    print("Validation failed:")
    print(e)
```

```python
# Create an invalid ticket response
invalid_response = {
    "category": "Internet Problem",
    "summary": "The student's Wi-Fi keeps disconnecting."
}

# Try to validate the response
try:
    result = TicketResponse(**invalid_response)
    print("Valid response!")
    print(result.model_dump())

except ValidationError as e:
    print("Validation failed!")
    print(e)
```

```python
# Create a response with a very short summary
short_response = {
    "category": "Wi-Fi & Network",
    "summary": "Bad"
}

# Validate the response
try:
    result = TicketResponse(**short_response)
    print("Valid response!")
    print(result.model_dump())

except ValidationError as e:
    print("Validation failed!")
    print(e)
```

### JSON and Pydantic Validation

Pydantic was used to validate the ticket assistant's response.

A `TicketResponse` model was created with two fields:

- **category:** Must match one of the 10 predefined IT support categories.
- **summary:** Must be a string containing between 5 and 200 characters.

**Validation Tests:**

1. A valid response containing `Wi-Fi & Network` was accepted.
2. An invalid category, `Internet Problem`, was rejected.
3. A summary containing only `Bad` was rejected because it was shorter than the minimum length.

**Conclusion:** Pydantic successfully validates the response structure and prevents unsupported categories and summaries that do not meet the defined length requirements.

```python
def classify_ticket(ticket):

    # Check whether the ticket is empty
    if not ticket or not ticket.strip():
        raise ValueError("Ticket cannot be empty.")

    # Predict category using keyword baseline
    category = keyword_baseline(ticket)

    # Generate a simple one-sentence summary
    summary = f"The student reports: {ticket.strip()}"

    # Prepare the response
    response = {
        "category": category,
        "summary": summary
    }

    # Validate response using Pydantic
    validated_response = TicketResponse(**response)

    # Return validated response as a dictionary
    return validated_response.model_dump()
```

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field

print("FastAPI libraries imported successfully!")
```

```python
# Create the FastAPI application
app = FastAPI(
    title="University IT Help Desk API",
    description="API for classifying student IT support tickets",
    version="1.0"
)

print("FastAPI application created successfully!")
```

```python
# Define the structure of incoming ticket data
class TicketRequest(BaseModel):
    ticket: str = Field(min_length=1, max_length=1000)

print("Ticket request model created successfully!")
```

```python
# Create the analyze endpoint
@app.post("/analyze")
def analyze_ticket(request: TicketRequest):

    try:
        # Send the ticket to our classification service
        result = classify_ticket(request.ticket)

        # Return the classification result
        return result

    except ValueError as e:
        raise HTTPException(
            status_code=400,
            detail=str(e)
        )

    except Exception:
        raise HTTPException(
            status_code=500,
            detail="Internal server error"
        )
```

```python
%%writefile service.py

from pydantic import BaseModel, Field
from typing import Literal

# Define the 10 allowed categories
Category = Literal[
    "Login & Account Access",
    "Wi-Fi & Network",
    "Software Installation",
    "Hardware Problem",
    "Email Issue",
    "LMS & Portal Issue",
    "Printer Issue",
    "Data & File Access",
    "Security & Phishing",
    "General IT Support"
]

# Validate the response
class TicketResponse(BaseModel):
    category: Category
    summary: str = Field(min_length=5, max_length=300)


# Classification service
def classify_ticket(ticket):

    if not ticket or not ticket.strip():
        raise ValueError("Ticket cannot be empty.")

    category = keyword_baseline(ticket)

    summary = f"The student reports: {ticket.strip()}"

    response = {
        "category": category,
        "summary": summary
    }

    validated_response = TicketResponse(**response)

    return validated_response.model_dump()
```

```python
%%writefile api.py

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field
from service import classify_ticket

app = FastAPI(
    title="University IT Help Desk API",
    description="API for classifying student IT support tickets",
    version="1.0"
)

class TicketRequest(BaseModel):
    ticket: str = Field(min_length=1, max_length=1000)

@app.post("/analyze")
def analyze_ticket(request: TicketRequest):
    try:
        result = classify_ticket(request.ticket)
        return result

    except ValueError as e:
        raise HTTPException(status_code=400, detail=str(e))

    except Exception:
        raise HTTPException(
            status_code=500,
            detail="Internal server error"
        )
```

```python
%%writefile frontend.py

import gradio as gr
import requests

API_URL = "http://127.0.0.1:8000/analyze"

def frontend_analyze(ticket):

    if not ticket or not ticket.strip():
        return "Error", "Please enter a ticket."

    try:
        response = requests.post(
            API_URL,
            json={"ticket": ticket},
            timeout=10
        )

        if response.status_code == 200:
            result = response.json()
            return result["category"], result["summary"]

        return "Error", response.json().get("detail", "Request failed.")

    except requests.RequestException:
        return "Error", "Unable to connect to the API."


demo = gr.Interface(
    fn=frontend_analyze,
    inputs=gr.Textbox(
        lines=4,
        label="Enter Student IT Ticket"
    ),
    outputs=[
        gr.Textbox(label="Predicted Category"),
        gr.Textbox(label="Ticket Summary")
    ],
    title="University IT Help Desk Assistant",
    description="Submit a student IT support ticket for classification.",
    submit_btn="Analyze Ticket"
)

if __name__ == "__main__":
    demo.launch()
```

```python
%%writefile service.py

from pydantic import BaseModel, Field
from typing import Literal

# Allowed categories
Category = Literal[
    "Login & Account Access",
    "Wi-Fi & Network",
    "Software Installation",
    "Hardware Problem",
    "Email Issue",
    "LMS & Portal Issue",
    "Printer Issue",
    "Data & File Access",
    "Security & Phishing",
    "General IT Support"
]

# Keyword rules
keyword_rules = {
    "Login & Account Access": [
        "login", "log in", "password", "username", "account",
        "sign in", "verification code"
    ],
    "Wi-Fi & Network": [
        "wifi", "wi-fi", "internet", "network", "connection",
        "disconnecting", "wireless", "wi fi"
    ],
    "Software Installation": [
        "install", "installation", "software", "update",
        "updating", "application", "setup"
    ],
    "Hardware Problem": [
        "laptop", "keyboard", "battery", "screen", "monitor",
        "computer", "touchpad", "usb ports"
    ],
    "Email Issue": [
        "email", "inbox", "outbox", "messages", "send emails",
        "receiving emails", "attachments"
    ],
    "LMS & Portal Issue": [
        "lms", "portal", "course page", "online quiz",
        "timetable", "attendance", "course materials"
    ],
    "Printer Issue": [
        "printer", "printing", "print", "paper jam",
        "printer queue", "blank pages"
    ],
    "Data & File Access": [
        "file", "files", "folder", "shared drive", "document",
        "deleted", "cloud storage", "access denied"
    ],
    "Security & Phishing": [
        "suspicious", "phishing", "virus", "antivirus",
        "unauthorized", "compromised", "harmful", "security",
        "unfamiliar login"
    ],
    "General IT Support": [
        "projector", "workstation", "technical assistance",
        "computer facilities", "it services",
        "technology facilities", "technical question"
    ]
}

# Keyword classification
def keyword_baseline(ticket):
    ticket = ticket.lower()

    security_keywords = [
        "suspicious", "phishing", "unauthorized",
        "compromised", "virus", "harmful"
    ]

    for keyword in security_keywords:
        if keyword in ticket:
            return "Security & Phishing"

    scores = {}

    for category, keywords in keyword_rules.items():
        score = 0

        for keyword in keywords:
            if keyword in ticket:
                score += 1

        scores[category] = score

    best_category = max(scores, key=scores.get)

    if scores[best_category] == 0:
        return "General IT Support"

    return best_category


# Response validation
class TicketResponse(BaseModel):
    category: Category
    summary: str = Field(min_length=5, max_length=300)


# Main classification service
def classify_ticket(ticket):

    if not ticket or not ticket.strip():
        raise ValueError("Ticket cannot be empty.")

    category = keyword_baseline(ticket)

    summary = f"The student reports: {ticket.strip()}"

    response = {
        "category": category,
        "summary": summary
    }

    validated_response = TicketResponse(**response)

    return validated_response.model_dump()
```

```python
import sys

sys.path.insert(
    0,
    "/content/University_IT_Help_Desk/source_code"
)

import gradio as gr
from fastapi.testclient import TestClient
import api

client = TestClient(api.app)

def analyze_ticket(ticket):
    response = client.post(
        "/analyze",
        json={"ticket": ticket}
    )

    if response.status_code == 200:
        result = response.json()

        return (
            "SUCCESS",
            result["category"],
            result["summary"]
        )

    return (
        "ERROR",
        "Not classified",
        str(response.json())
    )

with gr.Blocks() as app:

    gr.Markdown("# University IT Help Desk Assistant")
    gr.Markdown("Enter your university IT problem below.")

    ticket = gr.Textbox(
        label="Student Ticket",
        value="My university Wi-Fi is not working.",
        lines=3
    )

    button = gr.Button("Analyze Ticket")

    status = gr.Textbox(label="Status")
    category = gr.Textbox(label="Predicted Category")
    summary = gr.Textbox(label="Summary")

    button.click(
        analyze_ticket,
        inputs=ticket,
        outputs=[status, category, summary]
    )

app.launch(share=True)
```

```python
response = client.post(
    "/analyze",
    json={"ticket": "My Wi-Fi is not working."}
)

print("Normal API Status:", response.status_code)

# Simulate backend failure
def simulate_failure(ticket):
    raise RuntimeError("Backend unavailable")

original_function = api.classify_ticket
api.classify_ticket = simulate_failure

try:
    response = client.post(
        "/analyze",
        json={"ticket": "My Wi-Fi is not working."}
    )

    print("Failure Test Status:", response.status_code)
    print("Failure Details:", response.text)

finally:
    api.classify_ticket = original_function
```

```python
import os
import zipfile

project_path = "/content/University_IT_Help_Desk"
zip_path = "/content/University_IT_Help_Desk_Submission.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for root, dirs, files in os.walk(project_path):
        dirs[:] = [
            d for d in dirs
            if d not in ["__pycache__", ".pytest_cache"]
        ]

        for file in files:
            full_path = os.path.join(root, file)
            arcname = os.path.relpath(full_path, "/content")
            z.write(full_path, arcname)

print("FINAL ZIP CREATED SUCCESSFULLY!")
print(zip_path)
```
